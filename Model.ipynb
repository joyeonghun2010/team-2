```python
# 필요한 도구 불러오기
# =======================================================================
from sklearn.tree import DecisionTreeClassifier # 와인 종류를 분류할 트리 모델
from sklearn.model_selection import train_test_split # 학습/테스트 데이터 분리
from sklearn.metrics import accuracy_score # 맞힌 비율 계산
import pandas as pd # CSV 파일 읽기
import matplotlib.pyplot as plt # 그래프 그리기
# =======================================================================
```

```python
# 데이터 읽기: 성분을 입력값(X), 와인 종류를 정답(y)으로 분리
# =======================================================================
df = pd.read_csv("Wine dataset.csv") # CSV를 표로 읽기
X = df.drop(columns=['class']) # class 컬럼은 빼고, 나머지 성분 컬럼만 모델 입력으로 사용
y = df['class'] # 모델이 예측해야 할 와인 종류
# =======================================================================
```

```python
# 데이터 분할: 일부로 학습하고, 따로 남겨 둔 데이터로 모델 테스트
# =======================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2, # 전체의 20%는 테스트용으로 분할
    shuffle=True, # 분할 전에 데이터 순서를 섞음
    stratify=y, # 각 와인 종류의 비율이 학습/테스트에 비슷하게 들어가도록 설정
    random_state=42 # 같은 분할을 다시 만들 수 있게 시드 고정
)
# =======================================================================
```

```python
# 트리 설정을 직접 정하고 학습
# =======================================================================
clf = DecisionTreeClassifier(
    min_samples_leaf=2, # 마지막 분류에서 와인이 최소 2개는 들어가게 함
    random_state=42 # 같은 데이터에서 실행할 때 결과가 재현되도록 고정
)
clf = clf.fit(X_train, y_train) # 학습 데이터의 성분과 종류 관계를 학습
# =======================================================================
```

```python
# 예측: 테스트용 성분을 모델에 넣어 와인 종류를 맞혀 봄
# =======================================================================
pred = clf.predict(X_test) # 테스트용 성분을 모델에 넣어 와인 종류를 예측
print(pred) # 예측한 와인 종류 목록 출력
# =======================================================================
```

```python
# 최종 정확도: 테스트 정답 중 모델이 맞힌 비율
# =======================================================================
accuracy = accuracy_score(y_test, pred) # 맞힌 개수 / 전체 테스트 개수
print(f"정확도: {accuracy:.2%}") # 예: 100.00%는 테스트 표본을 모두 맞혔다는 뜻
# =======================================================================
```
